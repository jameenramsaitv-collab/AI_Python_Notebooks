```python
def celsius_to_fahrenheit(celsius):
    fahrenheit = []

    for temp in celsius:
        f = (temp * 9/5) + 32
        fahrenheit.append(f)

    return fahrenheit
```

```python
temperatures = [0, 10, 20, 25, 30, 100]

result = celsius_to_fahrenheit(temperatures)
```

```python

print("Celsius:", temperatures)
print("Fahrenheit:", result)
```
